```python
from __future__ import annotations

import re
import unicodedata
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.patches as patches
import pandas as pd
from openpyxl import load_workbook

# =========================================================
# CONFIGURACION
# =========================================================
INPUT_FILE = Path("/Users/martingomez/Downloads/Copia_BBDD_PowerBi.xlsx")
OUTPUT_FILE = Path("modelo_relacional_powerbi.xlsx")
DIAGRAM_FILE = Path("diagrama_modelo_relacional.png")

# =========================================================
# HELPERS
# =========================================================
def normalize_column(name: str) -> str:
    name = str(name).strip()
    name = unicodedata.normalize("NFKD", name).encode("ascii", "ignore").decode("ascii")
    name = re.sub(r"[^A-Za-z0-9]+", "_", name)
    name = re.sub(r"_+", "_", name).strip("_").lower()
    return name

def read_real_sheet(xlsx_path: Path, sheet_name: str, blank_streak_stop: int = 100) -> pd.DataFrame:
    wb = load_workbook(xlsx_path, read_only=True, data_only=True)
    ws = wb[sheet_name]

    rows = []
    header = None
    blank_streak = 0

    for row in ws.iter_rows(values_only=True):
        if header is None:
            header = list(row)
            continue

        is_blank = True
        for v in row:
            if v is None:
                continue
            if isinstance(v, str) and v.strip() == "":
                continue
            is_blank = False
            break

        if is_blank:
            blank_streak += 1
            if blank_streak >= blank_streak_stop:
                break
            continue

        blank_streak = 0
        rows.append(list(row))

    wb.close()
    return pd.DataFrame(rows, columns=header)

def clean_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [normalize_column(c) for c in df.columns]

    for c in df.columns:
        if df[c].dtype == "object":
            df[c] = df[c].astype(str).str.strip()
            df[c] = df[c].replace({
                "": pd.NA,
                "None": pd.NA,
                "nan": pd.NA,
                "NaT": pd.NA
            })
    return df

def add_surrogate_key(df: pd.DataFrame, key_name: str) -> pd.DataFrame:
    df = df.copy().reset_index(drop=True)
    df.insert(0, key_name, range(1, len(df) + 1))
    return df

def standardize_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    alias_map = {
        "no_pers": "n_pers",
        "n_pers": "n_pers",
        "n_pers_": "n_pers",
        "rut2": "rut2",
        "nombre": "nombre",
        "sexo": "sexo",
        "f_nacimiento": "f_nacimiento",
        "edad": "edad",
        "f_ingreso": "f_ingreso",
        "mes_de_antiguedad": "mes_de_antiguedad",
        "id_unidorg": "id_unidorg",
        "unidad_organizativa": "unidad_organizativa",
        "id_posic": "id_posic",
        "funcion": "funcion",
        "tipo_de_cargo_merco_talento": "tipo_de_cargo_merco_talento",
        "tipo_de_cargo": "tipo_de_cargo",
        "grupo": "grupo",
        "apers": "apers",
        "nombre_jefe": "nombre_jefe",
        "hsemanales": "hsemanales",
        "jornada": "jornada",
        "grper": "grper",
        "grupo_de_personal": "grupo_de_personal",
        "area_de_convenio": "area_de_convenio",
        "subtipo": "subtipo",
        "cc_nomina": "cc_nomina",
        "division_de_personal": "division_de_personal",
        "cecoste": "cecoste",
        "centro_de_coste": "centro_de_coste",
        "sociedad": "sociedad",
        "subdivision_de_personal": "subdivision_de_personal",
        "grprof": "grprof",
    }

    new_cols = {}
    for c in df.columns:
        if c in alias_map:
            new_cols[c] = alias_map[c]

    return df.rename(columns=new_cols)

def require_columns(df: pd.DataFrame, cols: list[str], df_name: str):
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise ValueError(
            f"En {df_name} faltan columnas requeridas: {missing}\n"
            f"Columnas disponibles: {list(df.columns)}"
        )

def to_nullable_numeric(df: pd.DataFrame, cols: list[str], force_int: bool = False) -> pd.DataFrame:
    df = df.copy()
    for c in cols:
        if c in df.columns:
            s = pd.to_numeric(df[c], errors="coerce")

            if force_int:
                # solo convierte a entero si realmente todos los no nulos son enteros
                mask_notna = s.notna()
                if ((s[mask_notna] % 1) == 0).all():
                    df[c] = s.astype("Int64")
                else:
                    df[c] = s.astype("Float64")
            else:
                df[c] = s.astype("Float64")
    return df

def to_string_key(df: pd.DataFrame, cols: list[str]) -> pd.DataFrame:
    """
    Convierte llaves a texto estandar para merge seguro.
    Ej: 1001, '1001 ', '1001.0' -> '1001'
    """
    df = df.copy()
    for c in cols:
        if c in df.columns:
            s = df[c]

            # Primero intentar numerico
            s_num = pd.to_numeric(s, errors="coerce")

            s_clean = s_num.astype("Int64").astype("string")
            # Si habia valores no numericos, conservarlos como string limpio
            mask_no_num = s_num.isna() & s.notna()
            if mask_no_num.any():
                s_alt = s.astype("string").str.strip()
                s_alt = s_alt.str.replace(r"\.0$", "", regex=True)
                s_clean = s_clean.mask(mask_no_num, s_alt)

            df[c] = s_clean
    return df

# =========================================================
# EXTRACCION
# =========================================================
empleado_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Empleado")))
cargo_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Cargo")))
unidad_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Unidad_Organizacional")))
jefatura_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Jefatura")))

wb_temp = load_workbook(INPUT_FILE, read_only=True)
sheetnames = wb_temp.sheetnames
wb_temp.close()

if "Condicion_Laboral" in sheetnames:
    condicion_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Condicion_Laboral")))
else:
    condicion_raw = standardize_columns(clean_dataframe(read_real_sheet(INPUT_FILE, "Condición_Laboral")))

# =========================================================
# VALIDACION
# =========================================================
require_columns(
    empleado_raw,
    ["n_pers", "rut2", "nombre", "sexo", "f_nacimiento", "edad", "f_ingreso", "mes_de_antiguedad"],
    "Empleado"
)

require_columns(
    cargo_raw,
    ["id_posic", "funcion", "tipo_de_cargo_merco_talento", "tipo_de_cargo", "grupo", "apers"],
    "Cargo"
)

require_columns(
    unidad_raw,
    ["id_unidorg", "unidad_organizativa"],
    "Unidad_Organizacional"
)

require_columns(
    condicion_raw,
    ["hsemanales", "jornada", "grper", "grupo_de_personal", "area_de_convenio", "subtipo", "cc_nomina"],
    "Condicion_Laboral"
)

require_columns(
    jefatura_raw,
    [
        "n_pers", "rut2", "id_unidorg", "id_posic", "nombre_jefe",
        "division_de_personal", "cecoste", "centro_de_coste",
        "sociedad", "subdivision_de_personal", "grprof",
        "edad", "mes_de_antiguedad", "hsemanales",
        "jornada", "grper", "grupo_de_personal",
        "area_de_convenio", "subtipo", "cc_nomina"
    ],
    "Jefatura"
)

# =========================================================
# TIPOS DE DATOS
# =========================================================
key_cols = ["n_pers", "rut2", "id_unidorg", "id_posic"]

empleado_raw = to_string_key(empleado_raw, key_cols)
cargo_raw = to_string_key(cargo_raw, key_cols)
unidad_raw = to_string_key(unidad_raw, key_cols)
jefatura_raw = to_string_key(jefatura_raw, key_cols)
condicion_raw = to_string_key(condicion_raw, [])  # no aplica

numeric_cols_common = ["edad", "mes_de_antiguedad", "hsemanales"]

empleado_raw = to_nullable_numeric(empleado_raw, numeric_cols_common, force_int=False)
jefatura_raw = to_nullable_numeric(jefatura_raw, numeric_cols_common, force_int=False)
condicion_raw = to_nullable_numeric(condicion_raw, ["hsemanales"], force_int=False)

empleado_raw["f_nacimiento"] = pd.to_datetime(empleado_raw["f_nacimiento"], errors="coerce")
empleado_raw["f_ingreso"] = pd.to_datetime(empleado_raw["f_ingreso"], errors="coerce")

# =========================================================
# DIMENSION EMPLEADO
# =========================================================
dim_empleado = (
    empleado_raw[
        [
            "n_pers",
            "rut2",
            "nombre",
            "sexo",
            "f_nacimiento",
            "edad",
            "f_ingreso",
            "mes_de_antiguedad"
        ]
    ]
    .dropna(subset=["n_pers", "rut2"])
    .drop_duplicates()
    .sort_values(["n_pers", "rut2"])
    .reset_index(drop=True)
)
dim_empleado = add_surrogate_key(dim_empleado, "empleado_sk")

# =========================================================
# DIMENSION CARGO
# =========================================================
dim_cargo = (
    cargo_raw[
        [
            "id_posic",
            "funcion",
            "tipo_de_cargo_merco_talento",
            "tipo_de_cargo",
            "grupo",
            "apers"
        ]
    ]
    .dropna(subset=["id_posic"])
    .drop_duplicates()
    .sort_values("id_posic")
    .reset_index(drop=True)
)
dim_cargo = add_surrogate_key(dim_cargo, "cargo_sk")

# =========================================================
# DIMENSION UNIDAD ORGANIZACIONAL
# =========================================================
dim_unidad_organizacional = (
    unidad_raw[
        [
            "id_unidorg",
            "unidad_organizativa"
        ]
    ]
    .dropna(subset=["id_unidorg"])
    .drop_duplicates()
    .sort_values("id_unidorg")
    .reset_index(drop=True)
)
dim_unidad_organizacional = add_surrogate_key(
    dim_unidad_organizacional,
    "unidad_organizacional_sk"
)

# =========================================================
# DIMENSION CONDICION LABORAL
# =========================================================
cond_cols = [
    "hsemanales",
    "jornada",
    "grper",
    "grupo_de_personal",
    "area_de_convenio",
    "subtipo",
    "cc_nomina"
]

dim_condicion_laboral = (
    condicion_raw[cond_cols]
    .dropna(how="all")
    .drop_duplicates()
    .sort_values(cond_cols, na_position="last")
    .reset_index(drop=True)
)
dim_condicion_laboral = add_surrogate_key(
    dim_condicion_laboral,
    "condicion_laboral_sk"
)

# =========================================================
# DIMENSION JEFATURA
# =========================================================
dim_jefatura = (
    jefatura_raw[
        [
            "nombre_jefe"
        ]
    ]
    .dropna(subset=["nombre_jefe"])
    .drop_duplicates()
    .sort_values("nombre_jefe")
    .reset_index(drop=True)
)
dim_jefatura = add_surrogate_key(dim_jefatura, "jefatura_sk")

# =========================================================
# TABLA DE HECHOS
# =========================================================
fact_dotacion = jefatura_raw.copy()

fact_dotacion = fact_dotacion.dropna(
    subset=["n_pers", "rut2", "id_unidorg", "id_posic"]
)

fact_dotacion = fact_dotacion.merge(
    dim_empleado[["empleado_sk", "n_pers", "rut2"]],
    on=["n_pers", "rut2"],
    how="left"
)

fact_dotacion = fact_dotacion.merge(
    dim_cargo[["cargo_sk", "id_posic"]],
    on="id_posic",
    how="left"
)

fact_dotacion = fact_dotacion.merge(
    dim_unidad_organizacional[
        ["unidad_organizacional_sk", "id_unidorg"]
    ],
    on="id_unidorg",
    how="left"
)

fact_dotacion = fact_dotacion.merge(
    dim_condicion_laboral[
        ["condicion_laboral_sk"] + cond_cols
    ],
    on=cond_cols,
    how="left"
)

fact_dotacion = fact_dotacion.merge(
    dim_jefatura[
        ["jefatura_sk", "nombre_jefe"]
    ],
    on="nombre_jefe",
    how="left"
)

fact_dotacion = fact_dotacion[
    [
        "empleado_sk",
        "cargo_sk",
        "unidad_organizacional_sk",
        "condicion_laboral_sk",
        "jefatura_sk",
        "division_de_personal",
        "cecoste",
        "centro_de_coste",
        "sociedad",
        "subdivision_de_personal",
        "grprof",
        "edad",
        "mes_de_antiguedad",
        "hsemanales"
    ]
].copy()

fact_dotacion.insert(0, "fact_dotacion_sk", range(1, len(fact_dotacion) + 1))

# =========================================================
# DICCIONARIO / RESUMEN
# =========================================================
diccionario = pd.DataFrame(
    [
        ["dim_empleado", "empleado_sk", "PK", "Clave sustituta del empleado"],
        ["dim_empleado", "n_pers", "NK", "Identificador natural del empleado"],
        ["dim_cargo", "cargo_sk", "PK", "Clave sustituta del cargo"],
        ["dim_cargo", "id_posic", "NK", "Identificador natural del cargo"],
        ["dim_unidad_organizacional", "unidad_organizacional_sk", "PK", "Clave sustituta de unidad"],
        ["dim_unidad_organizacional", "id_unidorg", "NK", "Identificador natural de unidad"],
        ["dim_condicion_laboral", "condicion_laboral_sk", "PK", "Clave sustituta de condicion laboral"],
        ["dim_jefatura", "jefatura_sk", "PK", "Clave sustituta de jefatura"],
        ["fact_dotacion", "fact_dotacion_sk", "PK", "Clave sustituta de tabla de hechos"],
        ["fact_dotacion", "empleado_sk", "FK", "Relacion con dim_empleado"],
        ["fact_dotacion", "cargo_sk", "FK", "Relacion con dim_cargo"],
        ["fact_dotacion", "unidad_organizacional_sk", "FK", "Relacion con dim_unidad_organizacional"],
        ["fact_dotacion", "condicion_laboral_sk", "FK", "Relacion con dim_condicion_laboral"],
        ["fact_dotacion", "jefatura_sk", "FK", "Relacion con dim_jefatura"]
    ],
    columns=["tabla", "columna", "tipo", "descripcion"]
)

resumen = pd.DataFrame(
    [
        ["dim_empleado", len(dim_empleado)],
        ["dim_cargo", len(dim_cargo)],
        ["dim_unidad_organizacional", len(dim_unidad_organizacional)],
        ["dim_condicion_laboral", len(dim_condicion_laboral)],
        ["dim_jefatura", len(dim_jefatura)],
        ["fact_dotacion", len(fact_dotacion)]
    ],
    columns=["tabla", "filas"]
)

# =========================================================
# EXPORTAR
# =========================================================
with pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl") as writer:
    dim_empleado.to_excel(writer, sheet_name="dim_empleado", index=False)
    dim_cargo.to_excel(writer, sheet_name="dim_cargo", index=False)
    dim_unidad_organizacional.to_excel(writer, sheet_name="dim_unidad_org", index=False)
    dim_condicion_laboral.to_excel(writer, sheet_name="dim_condicion_laboral", index=False)
    dim_jefatura.to_excel(writer, sheet_name="dim_jefatura", index=False)
    fact_dotacion.to_excel(writer, sheet_name="fact_dotacion", index=False)
    diccionario.to_excel(writer, sheet_name="diccionario_modelo", index=False)
    resumen.to_excel(writer, sheet_name="resumen", index=False)

print("Archivo generado:", OUTPUT_FILE)
print(resumen)

# =========================================================
# DIAGRAMA RELACIONAL
# =========================================================
fig, ax = plt.subplots(figsize=(16, 9))
ax.set_xlim(0, 16)
ax.set_ylim(0, 9)
ax.axis("off")

boxes = {
    "dim_empleado": (0.8, 5.8, 3.2, 2.3),
    "dim_cargo": (0.8, 2.8, 3.2, 2.0),
    "dim_unidad_org": (12.0, 5.9, 3.0, 1.6),
    "dim_condicion_laboral": (11.4, 2.4, 3.6, 2.3),
    "dim_jefatura": (6.2, 6.4, 3.2, 1.2),
    "fact_dotacion": (5.4, 2.2, 4.4, 3.0)
}

for title, (x, y, w, h) in boxes.items():
    rect = patches.FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.03",
        linewidth=1.5,
        edgecolor="black",
        facecolor="lightgray"
    )
    ax.add_patch(rect)
    ax.text(x + w / 2, y + h / 2, title, ha="center", va="center", fontsize=11, fontweight="bold")

ax.annotate("", xy=(5.4, 4.8), xytext=(4.0, 6.6), arrowprops=dict(arrowstyle="->"))
ax.annotate("", xy=(5.4, 3.5), xytext=(4.0, 3.8), arrowprops=dict(arrowstyle="->"))
ax.annotate("", xy=(9.8, 4.8), xytext=(12.0, 6.6), arrowprops=dict(arrowstyle="->"))
ax.annotate("", xy=(9.8, 3.5), xytext=(11.4, 3.5), arrowprops=dict(arrowstyle="->"))
ax.annotate("", xy=(8.0, 5.2), xytext=(8.0, 6.4), arrowprops=dict(arrowstyle="->"))

plt.title("Modelo Relacional Power BI", fontsize=16)
plt.savefig(DIAGRAM_FILE, dpi=300, bbox_inches="tight")
plt.show()
```
